# **FinTrust Data Exploration & Quality Audit**
This notebook performs exploratory data analysis (EDA) on FinTrust's customer and transaction datasets — inspecting structure, checking for missing values, and validating data integrity as a foundation for deeper financial and behavioral analysis.

```python
import pandas as pd
```

**2. Define File Paths.**
Storing the file paths as variables (`Path1`, `Path2`) keeps the code clean and makes it easy to update the source files later without editing multiple lines.***

```python
Path1 = "FinTrust_CustomerData.csv"      
```

```python
Path2 = "FinTrust_TransactionData.csv"   
```

**3. Load Customer Data**
We load `FinTrust_CustomerData.csv` into a DataFrame (`df1`) and preview the first 5 rows with `.head()`. This dataset holds customer-level information — demographics, account type, tenure, engagement score, and account status..**

```python
df1 = pd.read_csv(Path1)
df1.head()
```

**4. Inspect Customer Data Structure**
Using `.shape` and `.info()` to check the dataset's size, column names, and data types. This confirms we have **1,500 customers** across **12 columns**, with no missing values at a glance.

```python
df1.shape
df1.info()
```

**5. Load Transaction Data**
We load `FinTrust_TransactionData.csv` into a second DataFrame (`df2`). This dataset captures individual transaction records — type, amount, channel, device, location, and risk flags.

```python
df2 = pd.read_csv(Path2)
df2.head()
```

**6. Inspect Transaction Data Structure**
Checking size and data types confirms **12,000 transactions** across **11 columns**. Notably, `Device_Type` and `Location` show fewer non-null entries than the rest — a sign of missing data to investigate further.

```python
df2.shape
df2.info()
```

 **7. Identify Categorical vs Numerical Columns**
Separating columns by data type helps determine which fields are suited for statistical summaries (numerical) versus frequency/category analysis (categorical) later in the workflow.

```python
df1.select_dtypes(include='object').columns    # categorical
```

```python
df1.select_dtypes(include='number').columns    # numerical
```

```python
df2.select_dtypes(include='object').columns    # categorical
```

```python
df2.select_dtypes(include='number').columns    # numerical
```

 **8. Convert Transaction Timestamps**
Converting `Transaction_DateTime` from plain text into a proper datetime format. This enables time-based analysis — such as filtering by date range or extracting transaction hour/day patterns.

```python
df2['Transaction_DateTime'] = pd.to_datetime(df2['Transaction_DateTime'])
```

**9. Check for Missing Values**
Running `.isnull().sum()` on both datasets. The customer data (`df1`) is complete with zero missing values. The transaction data (`df2`) has **96 missing entries** in both `Device_Type` and `Location`, flagged for further cleaning.

```python
df1.isnull().sum()  #missing values
```

```python
df2.isnull().sum()    #missing values
```

 **10. Validate Data Integrity Between Tables**
Checking whether every transaction links to a real customer (referential integrity) — result: **0 orphaned transactions**. We also examine how many transactions each customer made, finding an average of **8 transactions per customer** (range: 1–20).

```python
orphans = set(df1['Customer_ID']) - set(df2['Customer_ID'])
print("Transactions with unknown customer:", len(orphans))

df2_per_customer = df2.groupby('Customer_ID').size()
print(df2_per_customer.describe())
```

 **11. Flag Data Quality Issues**
Final checks identify transactions with invalid (zero or negative) amounts and confirm the **96 rows** missing `Device_Type` — both flagged as candidates for cleaning or exclusion before deeper analysis.

```python
(df2['Amount_NGN'] <= 0).sum()            #invalid amounts
df2[df2['Device_Type'].isnull()].shape    #rows with missing device
```
